# Load the Preprocessed Dataset

```python
dataset=pd.read_csv("sensor_failure_analysis.csv",nrows=20000)
```

```python
dataset.columns
```

```python
#dataset=dataset.drop(['Machine_ID','Laser_Intensity','Hydraulic_Pressure_bar','Coolant_Flow_L_min','Heat_Index'],axis=1,inplace=True)
```

```python
dataset=pd.get_dummies(dataset,drop_first=True,dtype=int)
```

```python
dataset
```

# Input and Output Columns Separation:

```python
indep=dataset[[ 'Operational_Hours', 'Temperature_C', 'Vibration_mms', 'Oil_Level_pct', 'Remaining_Useful_Life_days']]
indep
```

```python
dep=dataset[['Failure_Within_7_Days']]
dep
```

```python
dataset["Failure_Within_7_Days"].value_counts()
```

# Train and Test dataset Split:

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(indep,dep,test_size=1/3,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train_= sc.fit_transform(X_train)
X_test_= sc.transform(X_test)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
param_grid={'n_estimators':[10,50,100], 'criterion':['entropy','log_loss','gini'],
            'random_state':[0],'max_features':['sqrt','log2',None]}
grid=GridSearchCV(RandomForestClassifier(),param_grid,refit=True,verbose=3,n_jobs=-1)
grid.fit(X_train,y_train)
```

```python
re=grid.cv_results_

grid_predictions=grid.predict(X_test)

from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,grid_predictions)


from sklearn.metrics import classification_report
clf_report=classification_report(y_test,grid_predictions)

from sklearn.metrics import f1_score
f1_macro=f1_score(y_test,grid_predictions,average='weighted')
print("The f1_macro value for best parameter {}:".format(grid.best_params_),f1_macro)
```
